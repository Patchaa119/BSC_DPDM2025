```python
# ============================================================
# BaggingRegressor Rainfall Prediction - 1 Cluster = 1 Model
# ============================================================

import pandas as pd
import numpy as np
import os
import copy
import pickle
import warnings
from dateutil.relativedelta import relativedelta
from sklearn.ensemble import BaggingRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.multioutput import MultiOutputRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from scipy.stats import percentileofscore
import matplotlib.pyplot as plt
warnings.filterwarnings('ignore')

algo          = 'Bagging_V1'
window_size   = 24
horizon       = 12
k_fold_num    = 2
n_estimators  = 100
max_depth     = 4
max_samples   = 0.8
max_features  = 0.8                                    # ✅ New: fraction of features per base estimator

# ============================================================
# โหลดข้อมูล
# ============================================================
X_index    = pd.read_csv('/content/drive/MyDrive/KKU_Workshop3/X_variable(Index)_operation_v4.csv')
Y_rainfall = pd.read_csv('/content/drive/MyDrive/KKU_Workshop3/Y_variable(Rainfall).csv')
node       = pd.read_csv('/content/drive/MyDrive/KKU_Workshop3/Node_table_TMD.csv')
feature    = pd.read_csv('/content/drive/MyDrive/KKU_Workshop3/lookup_table_reanalysis_v5.csv', header='infer')

X_index['DATE'] = pd.to_datetime(X_index['DATE'])
X_index['DATE'] = X_index['DATE'].dt.strftime('%Y-%m-01')
Y_rainfall      = Y_rainfall.rename(columns={'code': 'Node'})
Y_rainfall_clean = Y_rainfall.copy()

cluster_list = sorted(feature['Cluster'].dropna().unique())
print(f"Cluster จาก lookup_table: {cluster_list}")

# แสดงสถานีในแต่ละ cluster
print(f"\n{'='*60}")
print("สถานีในแต่ละ Cluster")
print(f"{'='*60}")
for cluster in cluster_list:
    stations_in_cluster = [
        node['Node'].iloc[i]
        for i in range(len(node))
        if node['quality'].iloc[i] == 1 and
           node['cluster'].iloc[i] == cluster
    ]
    print(f"Cluster {int(cluster):2d} ({len(stations_in_cluster)} สถานี): {stations_in_cluster}")

# ============================================================
# เติมค่า Missing Value
# ============================================================
HQ_date_get_month_df = Y_rainfall.copy()
HQ_date_get_month_df['DATE'] = pd.to_datetime(Y_rainfall['DATE']).dt.month

month_avg_list = []
for month in range(1, 13):
    HQ_data    = HQ_date_get_month_df[HQ_date_get_month_df['DATE'] == month]
    av_r       = HQ_data.iloc[:, 1:].median()
    month_dict = {'MONTH': month}
    month_dict.update(av_r)
    month_avg_list.append(month_dict)
HQ_month_avg = pd.DataFrame(month_avg_list)

for i in range(1, len(Y_rainfall_clean.columns)):
    for j in range(len(Y_rainfall_clean)):
        if np.isnan(Y_rainfall_clean.iloc[j, i]):
            Y_rainfall_clean.iloc[j, i] = HQ_month_avg.iloc[
                pd.to_datetime(Y_rainfall_clean.iloc[:, 0]).dt.month[j]-1, i
            ]

# ============================================================
# Time Lag + Normalize
# ============================================================
def apply_time_lag(df, column_name, lag):
    df[f"{column_name}_lag{lag}"] = df[column_name].shift(lag)
    return df

time_lags = {
    'MEIV2': [1,2,3], 'RMM_AMPLITUDE': [1], 'PHASE': [1],
    'PDO': [1,2], 'ONI': [1,2], 'DMI': [1,2],
    'BSISO1': [1], 'BSISO1-Phase': [1], 'SWM': [1],
}
for column, lags in time_lags.items():
    if column in X_index.columns:
        for lag in lags:
            X_index = apply_time_lag(X_index, column, lag)

X_index['DATE']          = pd.to_datetime(X_index['DATE'])
Y_rainfall_clean['DATE'] = pd.to_datetime(Y_rainfall_clean['DATE'])

start_date        = Y_rainfall_clean['DATE'].min()
max_date_possible = min(X_index['DATE'].max(), Y_rainfall_clean['DATE'].max())

con_X = (X_index['DATE'] >= start_date) & (X_index['DATE'] <= max_date_possible)
con_Y = (Y_rainfall_clean['DATE'] >= start_date) & (Y_rainfall_clean['DATE'] <= max_date_possible)

X_index_interval_date    = X_index.loc[con_X, :]
Y_rainfall_clean         = Y_rainfall_clean.loc[con_Y, :]
Y_rainfall_interval_date = copy.deepcopy(Y_rainfall_clean)

def convert_columns_to_index_columns(df, col_name):
    df.loc[:, col_name] = pd.to_datetime(df[col_name])
    return df.set_index(col_name, inplace=False)

X_index_df_ready    = convert_columns_to_index_columns(X_index_interval_date, 'DATE')
Y_rainfall_df_ready = convert_columns_to_index_columns(Y_rainfall_interval_date, 'DATE')

scaler_X         = StandardScaler()
scaler_Y         = StandardScaler()
X_normalize_data = scaler_X.fit_transform(X_index_df_ready)
Y_normalize_data = scaler_Y.fit_transform(Y_rainfall_df_ready)

X_index_normalized_df = pd.DataFrame(
    X_normalize_data,
    columns=X_index_df_ready.columns,
    index=X_index_df_ready.index
)
Y_rainfall_normalized_df = pd.DataFrame(
    Y_normalize_data,
    columns=Y_rainfall_df_ready.columns,
    index=Y_rainfall_df_ready.index
)

with open('/content/drive/MyDrive/KKU_Workshop3/scaler_Y.pkl', 'wb') as f:
    pickle.dump(scaler_Y, f)
with open('/content/drive/MyDrive/KKU_Workshop3/scaler_X.pkl', 'wb') as f:
    pickle.dump(scaler_X, f)

# ✅ ตัด NaN และ align X กับ Y ให้ตรงกัน
X_index_normalized_df = X_index_normalized_df.dropna()
common_index = X_index_normalized_df.index.intersection(
    Y_rainfall_normalized_df.index
)
X_index_normalized_df    = X_index_normalized_df.loc[common_index]
Y_rainfall_normalized_df = Y_rainfall_normalized_df.loc[common_index]

X_index_normalized_df['Month_sin'] = np.sin(2 * np.pi * X_index_normalized_df.index.month / 12)
X_index_normalized_df['Month_cos'] = np.cos(2 * np.pi * X_index_normalized_df.index.month / 12)

n_feature_climate = X_index_normalized_df.shape[1]
feature_climate   = X_index_normalized_df.columns

print(f"\nX shape: {X_index_normalized_df.shape}")
print(f"Y shape: {Y_rainfall_normalized_df.shape}")
print(f"NaN in X: {X_index_normalized_df.isnull().sum().sum()}")
print(f"NaN in Y: {Y_rainfall_normalized_df.isnull().sum().sum()}")

# ============================================================
# ฟังก์ชันที่ใช้งาน
# ============================================================
def nse(observations, predictions):
    mean_obs    = np.mean(observations)
    numerator   = np.sum((observations - predictions) ** 2)
    denominator = np.sum((observations - mean_obs) ** 2)
    if denominator == 0:
        return float('nan')
    return float(1 - (numerator / denominator))


def quantile_mapper(value, historical_preds, historical_obs):
    percentile = percentileofscore(historical_preds, value)
    percentile = float(np.clip(percentile, 0, 100))  # ✅ clamp [0, 100]
    corrected_value = np.percentile(
        historical_obs, percentile, method='closest_observation'
    )
    return corrected_value


def get_selected_index(cluster):
    feature_cluster = feature[feature['Cluster'] == cluster]
    selected_index  = []
    for feature_index in range(len(feature_cluster)):
        for i in range(len(feature_climate)):
            if (feature_climate[i] == feature_cluster.iloc[feature_index, 1]) and \
               (feature_cluster.iloc[feature_index, 5] == 0):
                selected_index.append(i)
            elif feature_cluster.iloc[feature_index, 5] != 0:
                for i_lag in range(len(feature_climate)):
                    if feature_climate[i_lag] == \
                       feature_cluster.iloc[feature_index, 1] + \
                       '_lag' + str(feature_cluster.iloc[feature_index, 5]):
                        selected_index.append(i_lag)
                        break
                break
    selected_index.sort()
    return selected_index


def prepare_flat_data_per_station(df, window_size, horizon):
    X_all, y_all = [], []
    df = df.dropna()
    if len(df) < window_size + horizon:
        return np.array([]), np.array([])
    for j in range(len(df) - window_size - horizon + 1):
        x_window = df.iloc[j:j+window_size, :-1].values
        y_target = df.iloc[j+window_size:j+window_size+horizon, -1].values
        if np.isnan(x_window).any() or np.isnan(y_target).any():
            continue
        x_flat = x_window.flatten()
        x_mean = x_window.mean(axis=0)
        x_std  = x_window.std(axis=0)
        x_max  = x_window.max(axis=0)
        x_min  = x_window.min(axis=0)
        x_last = x_window[-1, :]
        month     = df.index[j + window_size].month
        month_sin = np.sin(2 * np.pi * month / 12)
        month_cos = np.cos(2 * np.pi * month / 12)
        x_combined = np.concatenate([
            x_flat, x_mean, x_std, x_max, x_min, x_last,
            [month_sin, month_cos]
        ])
        X_all.append(x_combined)
        y_all.append(y_target)
    return np.array(X_all), np.array(y_all)


def select_feature_f(list_df, position_feature_index):
    result = copy.deepcopy(list_df)
    for i in range(len(result)):
        result[i] = result[i].iloc[:, position_feature_index]
    return result

# ============================================================
# Loop หลัก 1 Cluster = 1 Model
# ============================================================
all_results = []

n     = len(Y_rainfall_normalized_df)
start = pd.to_datetime(min(Y_rainfall_normalized_df.index))

train_end_idx  = int(n * 0.70)
train_end_idx2 = int(n * 0.85)

end_train1  = start + relativedelta(months=train_end_idx)
start_val1  = start + relativedelta(months=train_end_idx - 12)
end_val1    = start + relativedelta(months=int(n * 0.85))
start_test1 = start + relativedelta(months=train_end_idx)
end_test1   = start + relativedelta(months=n)

end_train2  = start + relativedelta(months=train_end_idx2)
start_val2  = start + relativedelta(months=train_end_idx2 - 12)
end_val2    = start + relativedelta(months=n)
start_test2 = start + relativedelta(months=train_end_idx2)
end_test2   = start + relativedelta(months=n)

print(f"\nFold 1: Train {start} → {end_train1} | Val {start_val1} → {end_val1} | Test {start_test1} → {end_test1}")
print(f"Fold 2: Train {start} → {end_train2} | Val {start_val2} → {end_val2} | Test {start_test2} → {end_test2}")

for cluster in cluster_list:
    print(f"\n{'='*60}")
    print(f"CLUSTER {cluster}")
    print(f"{'='*60}")

    cluster_stations = [
        i for i in range(len(Y_rainfall_normalized_df.columns))
        if node['quality'].iloc[i] == 1 and
           node['cluster'].iloc[i] == cluster
    ]

    if len(cluster_stations) == 0:
        print(f"⚠️ ไม่มีสถานีใน Cluster {cluster} ข้ามไป")
        continue

    print(f"จำนวนสถานี: {len(cluster_stations)}")
    for s in cluster_stations:
        print(f"  - {node['Node'].iloc[s]}")

    num_feature    = len(feature[feature['Cluster'] == cluster])
    selected_index = get_selected_index(cluster)

    if num_feature == 0:
        print(f"⚠️ ไม่มี features ข้ามไป")
        continue

    position_optimal  = [0] * n_feature_climate
    position_remain   = [1]
    for idx in selected_index:
        position_optimal[idx] = 1
    position_all       = position_optimal + position_remain
    position_all_index = [i for i, v in enumerate(position_all) if v == 1]
    climate_name_list  = [feature_climate[i] for i, v in enumerate(position_optimal) if v == 1]

    print(f"Features ({num_feature}): {list(climate_name_list)}")

    list_table = []
    for code_st in Y_rainfall_normalized_df.columns:
        eindices_df          = X_index_normalized_df.copy()
        eindices_df[code_st] = Y_rainfall_normalized_df[code_st]
        list_table.append(eindices_df)

    list_table_filtered = select_feature_f(list_table, position_all_index)

    for fold_index in range(k_fold_num):
        print(f"\n--- Fold {fold_index + 1} ---")

        if fold_index == 0:
            train_list = [list_table_filtered[s][start:end_train1]     for s in cluster_stations]
            val_list   = [list_table_filtered[s][start_val1:end_val1]   for s in cluster_stations]
            test_list  = [list_table_filtered[s][start_test1:end_test1] for s in cluster_stations]
        else:
            train_list = [list_table_filtered[s][start:end_train2]     for s in cluster_stations]
            val_list   = [list_table_filtered[s][start_val2:end_val2]   for s in cluster_stations]
            test_list  = [list_table_filtered[s][start_test2:end_test2] for s in cluster_stations]

        X_train_all, y_train_all = [], []
        X_val_all,   y_val_all   = [], []
        X_test_all,  y_test_all  = [], []

        for i in range(len(cluster_stations)):
            X_tr, y_tr = prepare_flat_data_per_station(train_list[i], window_size, horizon)
            X_va, y_va = prepare_flat_data_per_station(val_list[i],   window_size, horizon)
            X_te, y_te = prepare_flat_data_per_station(test_list[i],  window_size, horizon)
            if len(X_tr) > 0: X_train_all.append(X_tr); y_train_all.append(y_tr)
            if len(X_va) > 0: X_val_all.append(X_va);   y_val_all.append(y_va)
            if len(X_te) > 0: X_test_all.append(X_te);  y_test_all.append(y_te)

        if len(X_train_all) == 0 or len(X_test_all) == 0 or len(X_val_all) == 0:
            print(f"⚠️ ข้อมูลไม่พอ ข้ามไป")
            continue

        X_train = np.vstack(X_train_all)
        y_train = np.vstack(y_train_all)
        X_val   = np.vstack(X_val_all)
        y_val   = np.vstack(y_val_all)
        X_test  = np.vstack(X_test_all)
        y_test  = np.vstack(y_test_all)

        print(f"Train={X_train.shape} | Val={X_val.shape} | Test={X_test.shape}")

        # ✅ Changed: AdaBoostRegressor → BaggingRegressor
        model = Pipeline([
            ('imputer', SimpleImputer(strategy='median')),
            ('model', MultiOutputRegressor(
                BaggingRegressor(
                    estimator=DecisionTreeRegressor(
                        max_depth=max_depth,
                        min_samples_split=5,
                        min_samples_leaf=3,
                        random_state=42
                    ),
                    n_estimators=n_estimators,
                    max_samples=max_samples,                 # ✅ fraction of training samples per tree
                    max_features=max_features,              # ✅ fraction of features per tree
                    bootstrap=True,                           # ✅ sample with replacement
                    bootstrap_features=False,                  # ✅ features without replacement
                    n_jobs=-1,
                    random_state=42
                ),
                n_jobs=-1
            ))
        ])

        print("Training...")
        model.fit(X_train, y_train)
        print("Done!")

        y_val_pred  = model.predict(X_val)
        y_test_pred = model.predict(X_test)

        # ✅ Quantile Mapping พร้อม clamp
        hist_preds = np.sort(y_val_pred.flatten())
        hist_obs   = np.sort(y_val.flatten())
        y_test_corrected = np.array([
            quantile_mapper(v, hist_preds, hist_obs)
            for v in y_test_pred.flatten()
        ]).reshape(y_test_pred.shape)

        val_r2    = r2_score(y_val, y_val_pred)
        val_nse   = nse(y_val.flatten(), y_val_pred.flatten())
        val_rmse  = np.sqrt(mean_squared_error(y_val, y_val_pred))
        test_r2   = r2_score(y_test, y_test_pred)
        test_nse  = nse(y_test.flatten(), y_test_pred.flatten())
        test_rmse = np.sqrt(mean_squared_error(y_test, y_test_pred))
        test_mse  = mean_squared_error(y_test, y_test_pred)
        qm_r2     = r2_score(y_test, y_test_corrected)
        qm_nse    = nse(y_test.flatten(), y_test_corrected.flatten())
        qm_rmse   = np.sqrt(mean_squared_error(y_test, y_test_corrected))

        acc_monthly = [
            r2_score(y_test[:, m], y_test_pred[:, m])
            for m in range(horizon)
        ]

        print(f"Val  → R²={val_r2:.4f}  NSE={val_nse:.4f}  RMSE={val_rmse:.4f}")
        print(f"Test → R²={test_r2:.4f}  NSE={test_nse:.4f}  RMSE={test_rmse:.4f}")
        print(f"QM   → R²={qm_r2:.4f}  NSE={qm_nse:.4f}  RMSE={qm_rmse:.4f}")

        station_names = [str(node['Node'].iloc[s]) for s in cluster_stations]

        all_results.append({
            'cluster': cluster,
            'stations': ', '.join(station_names),
            'n_stations': len(cluster_stations),
            'fold': fold_index,
            'num_features': num_feature,
            'features': ', '.join(list(climate_name_list)),
            'val_r2': val_r2,   'val_nse': val_nse,   'val_rmse': val_rmse,
            'test_r2': test_r2, 'test_nse': test_nse, 'test_rmse': test_rmse,
            'test_mse': test_mse,
            'qm_r2': qm_r2,    'qm_nse': qm_nse,     'qm_rmse': qm_rmse,
            **{f'r2_month{m+1}': acc_monthly[m] for m in range(horizon)}
        })

        fig, axes = plt.subplots(1, 2, figsize=(14, 4))
        axes[0].plot(y_test[:60, 0],           label='Actual',   linewidth=2)
        axes[0].plot(y_test_pred[:60, 0],      label='Bagging',  linewidth=2, linestyle='--')  # ✅ label updated
        axes[0].plot(y_test_corrected[:60, 0], label='QM',       linewidth=2, linestyle=':')
        axes[0].set_title(f"Cluster {cluster} Fold {fold_index+1} | NSE={test_nse:.4f}")
        axes[0].set_ylabel("Rainfall (normalized)")
        axes[0].legend()
        axes[0].grid(alpha=0.3)
        axes[1].bar(range(1, horizon+1), acc_monthly, color='steelblue')
        axes[1].set_title("R² Month")
        axes[1].set_xlabel("Month Ahead")
        axes[1].set_ylabel("R²")
        axes[1].grid(alpha=0.3)
        plt.tight_layout()
        plt.show()
```

```python
# ============================================================
# Cell 7: Training Loop — 1 Cluster = 1 Model (BaggingRegressor)
# ============================================================
import matplotlib.gridspec as gridspec


all_results = []


for cluster in cluster_list:
    print(f"\n{'='*60}")
    print(f"CLUSTER {cluster}")
    print(f"{'='*60}")

    cluster_stations = [
        i for i in range(len(Y_rainfall_normalized_df.columns))
        if node['quality'].iloc[i] == 1 and
           node['cluster'].iloc[i] == cluster
    ]

    if len(cluster_stations) == 0:
        print(f"⚠️ ไม่มีสถานีใน Cluster {cluster} ข้ามไป")
        continue

    print(f"จำนวนสถานี: {len(cluster_stations)}")
    for s in cluster_stations:
        print(f"  - {node['Node'].iloc[s]}")

    num_feature    = len(feature[feature['Cluster'] == cluster])
    selected_index = get_selected_index(cluster)

    if num_feature == 0:
        print(f"⚠️ ไม่มี features ข้ามไป")
        continue

    position_optimal  = [0] * n_feature_climate
    position_remain   = [1]
    for idx in selected_index:
        position_optimal[idx] = 1
    position_all       = position_optimal + position_remain
    position_all_index = [i for i, v in enumerate(position_all) if v == 1]
    climate_name_list  = [feature_climate[i] for i, v in enumerate(position_optimal) if v == 1]

    print(f"Features ({num_feature}): {list(climate_name_list)}")

    list_table = []
    for code_st in Y_rainfall_normalized_df.columns:
        eindices_df          = X_index_normalized_df.copy()
        eindices_df[code_st] = Y_rainfall_normalized_df[code_st]
        list_table.append(eindices_df)

    list_table_filtered = select_feature_f(list_table, position_all_index)

    for fold_index in range(k_fold_num):
        print(f"\n--- Fold {fold_index + 1} ---")

        if fold_index == 0:
            train_list = [list_table_filtered[s][start:end_train1]     for s in cluster_stations]
            val_list   = [list_table_filtered[s][start_val1:end_val1]   for s in cluster_stations]
            test_list  = [list_table_filtered[s][start_test1:end_test1] for s in cluster_stations]
        else:
            train_list = [list_table_filtered[s][start:end_train2]     for s in cluster_stations]
            val_list   = [list_table_filtered[s][start_val2:end_val2]   for s in cluster_stations]
            test_list  = [list_table_filtered[s][start_test2:end_test2] for s in cluster_stations]

        X_train_all, y_train_all = [], []
        X_val_all,   y_val_all   = [], []
        X_test_all,  y_test_all  = [], []

        for i in range(len(cluster_stations)):
            X_tr, y_tr = prepare_flat_data_per_station(train_list[i], window_size, horizon)
            X_va, y_va = prepare_flat_data_per_station(val_list[i],   window_size, horizon)
            X_te, y_te = prepare_flat_data_per_station(test_list[i],  window_size, horizon)
            if len(X_tr) > 0: X_train_all.append(X_tr); y_train_all.append(y_tr)
            if len(X_va) > 0: X_val_all.append(X_va);   y_val_all.append(y_va)
            if len(X_te) > 0: X_test_all.append(X_te);  y_test_all.append(y_te)

        if len(X_train_all) == 0 or len(X_test_all) == 0 or len(X_val_all) == 0:
            print(f"⚠️ ข้อมูลไม่พอ ข้ามไป")
            continue

        X_train = np.vstack(X_train_all)
        y_train = np.vstack(y_train_all)
        X_val   = np.vstack(X_val_all)
        y_val   = np.vstack(y_val_all)
        X_test  = np.vstack(X_test_all)
        y_test  = np.vstack(y_test_all)

        print(f"Train={X_train.shape} | Val={X_val.shape} | Test={X_test.shape}")

        # ✅ BaggingRegressor Model
        model = Pipeline([
            ('imputer', SimpleImputer(strategy='median')),
            ('model', MultiOutputRegressor(
                BaggingRegressor(
                    estimator=DecisionTreeRegressor(
                        max_depth=max_depth,
                        min_samples_split=5,
                        min_samples_leaf=3,
                        random_state=42
                    ),
                    n_estimators=n_estimators,
                    max_samples=max_samples,
                    max_features=max_features,
                    bootstrap=True,
                    bootstrap_features=False,
                    n_jobs=-1,
                    random_state=42
                ),
                n_jobs=-1
            ))
        ])

        print("Training...")
        model.fit(X_train, y_train)
        print("Done!")

        y_val_pred  = model.predict(X_val)
        y_test_pred = model.predict(X_test)

        hist_preds = np.sort(y_val_pred.flatten())
        hist_obs   = np.sort(y_val.flatten())
        y_test_corrected = np.array([
            quantile_mapper(v, hist_preds, hist_obs)
            for v in y_test_pred.flatten()
        ]).reshape(y_test_pred.shape)

        val_r2    = r2_score(y_val, y_val_pred)
        val_nse   = nse(y_val.flatten(), y_val_pred.flatten())
        val_rmse  = np.sqrt(mean_squared_error(y_val, y_val_pred))
        test_r2   = r2_score(y_test, y_test_pred)
        test_nse  = nse(y_test.flatten(), y_test_pred.flatten())
        test_rmse = np.sqrt(mean_squared_error(y_test, y_test_pred))
        test_mse  = mean_squared_error(y_test, y_test_pred)
        qm_r2     = r2_score(y_test, y_test_corrected)
        qm_nse    = nse(y_test.flatten(), y_test_corrected.flatten())
        qm_rmse   = np.sqrt(mean_squared_error(y_test, y_test_corrected))

        acc_monthly = [
            r2_score(y_test[:, m], y_test_pred[:, m])
            for m in range(horizon)
        ]

        print(f"Val  → R²={val_r2:.4f}  NSE={val_nse:.4f}  RMSE={val_rmse:.4f}")
        print(f"Test → R²={test_r2:.4f}  NSE={test_nse:.4f}  RMSE={test_rmse:.4f}")
        print(f"QM   → R²={qm_r2:.4f}  NSE={qm_nse:.4f}  RMSE={qm_rmse:.4f}")

        station_names = [str(node['Node'].iloc[s]) for s in cluster_stations]

        all_results.append({
            'cluster': cluster,
            'stations': ', '.join(station_names),
            'n_stations': len(cluster_stations),
            'fold': fold_index,
            'num_features': num_feature,
            'features': ', '.join(list(climate_name_list)),
            'val_r2': val_r2,   'val_nse': val_nse,   'val_rmse': val_rmse,
            'test_r2': test_r2, 'test_nse': test_nse, 'test_rmse': test_rmse,
            'test_mse': test_mse,
            'qm_r2': qm_r2,    'qm_nse': qm_nse,     'qm_rmse': qm_rmse,
            **{f'r2_month{m+1}': acc_monthly[m] for m in range(horizon)}
        })

        # ── Snapshot Plot ──────────────────────────────────────────────────────
        snapshot_t   = 0
        df_snap      = test_list[0].reset_index(drop=True)
        rain_col     = df_snap.columns[-1]
        clim_cols    = list(climate_name_list)
        station_name = str(node['Node'].iloc[cluster_stations[0]])

        obs_input       = df_snap[rain_col].iloc[snapshot_t : snapshot_t + window_size].values
        obs_target      = df_snap[rain_col].iloc[snapshot_t + window_size : snapshot_t + window_size + horizon].values
        pred_target     = y_test_pred[snapshot_t]
        df_climate_snap = df_snap[clim_cols].iloc[snapshot_t : snapshot_t + window_size + horizon]

        x_input  = np.arange(1, window_size + 1)
        x_target = np.arange(window_size + 1, window_size + horizon + 1)
        x_all    = np.arange(1, window_size + horizon + 1)

        fig = plt.figure(figsize=(13, 6))
        gs  = gridspec.GridSpec(2, 1, height_ratios=[2, 1], hspace=0.4)
        ax_top = fig.add_subplot(gs[0])
        ax_bot = fig.add_subplot(gs[1])

        for ax in [ax_top, ax_bot]:
            ax.axvline(x=window_size + 0.5, color='gray', linestyle=':', linewidth=1)

        ax_top.plot(x_input,  obs_input,   color='steelblue',  linewidth=2,
                    label=f'Observed Rainfall (input) ({window_size})')
        ax_top.plot(x_target, obs_target,  color='darkorange', linewidth=2,
                    linestyle='--', label=f'Observed Rainfall (target) ({horizon})')
        ax_top.plot(x_target, pred_target, color='green',      linewidth=2,
                    label=f'Rainfall prediction (BaggingRegressor)')  # ✅ Updated label

        ax_top.set_title(f'BaggingRegressor – Station {station_name} | Cluster {cluster} | Fold {fold_index+1} | snapshot t={snapshot_t}')  # ✅ Updated title
        ax_top.set_ylabel('Rainfall (normalise)')
        ax_top.legend(fontsize=9)
        ax_top.grid(alpha=0.3)
        ax_top.tick_params(labelbottom=False)
        ax_top.set_xlim(0.5, window_size + horizon + 0.5)

        for col in clim_cols:
            ax_bot.plot(x_all, df_climate_snap[col].values, linewidth=1.2, label=col)
        ax_bot.set_ylabel('Climate index')
        ax_bot.set_xlabel('Month')
        ax_bot.grid(alpha=0.3)
        ax_bot.set_xlim(0.5, window_size + horizon + 0.5)
        if len(clim_cols) <= 12:
            ax_bot.legend(fontsize=7, ncol=3, loc='upper left')

        plt.tight_layout()
        display(fig)
        plt.close(fig)
```
