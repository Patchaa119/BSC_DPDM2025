```python
import os
import re
import pandas as pd

daily_input_data = []

pattern = re.compile(r'^HII_daily_rainfall_\d{4}-\d{2}-\d{2}\.csv$')

monthly_folders = [
    f for f in os.listdir(input_path)
    if os.path.isdir(os.path.join(input_path, f))
]
monthly_folders.sort()

for month in monthly_folders:
    month_path = os.path.join(input_path, month)

    files = [
        f for f in os.listdir(month_path)
        if pattern.match(f)
    ]

    print(month, 'valid daily files:', len(files))  # sanity check

    for f in files:
        file_path = os.path.join(month_path, f)
        try:
            df = pd.read_csv(file_path)

            df['Source_Month'] = month
            df['Source_File'] = f

            daily_input_data.append(df)
        except pd.errors.EmptyDataError:
            print(f"Skipping empty file: {file_path}")
        except Exception as e:
            print(f"Error reading file {file_path}: {e}")

daily_input_df = pd.concat(daily_input_data, ignore_index=True)
```

# **ข้อ3** ระบุข้อมูลสูญหาย (record ที่หายไป) ของแต่ละสถานี แสดงจำนวนและสัดส่วน

```python
daily_input_df['date'] = pd.to_datetime(
    daily_input_df['rainfall_datetime_calc']
).dt.date
```

```python
station_list = station_df['Node'].unique()
all_dates = daily_input_df['date'].unique()
```

```python
expected = pd.MultiIndex.from_product(
    [station_list, all_dates],
    names=['tele_station_oldcode', 'date']
).to_frame(index=False)
```

```python
observed = daily_input_df[['tele_station_oldcode', 'date']]
```

```python
missing_daily = expected.merge(
    observed,
    on=['tele_station_oldcode', 'date'],
    how='left',
    indicator=True
)

missing_daily = missing_daily[missing_daily['_merge'] == 'left_only']
```

```python
missing_by_station = (
    missing_daily
    .groupby('tele_station_oldcode')
    .size()
    .reset_index(name='Missing_records')
)

total_days = len(all_dates)
missing_by_station['Total_records'] = total_days
missing_by_station['Missing_ratio'] = (
    missing_by_station['Missing_records'] / total_days
)
missing_by_station['Missing_percent'] = missing_by_station['Missing_ratio'] * 100
missing_by_station.sort_values('Missing_percent', ascending=False)
```

```python
excluded_stations = missing_by_station[
    missing_by_station['Missing_percent'] >= 20
]
```

```python
excluded_stations[['tele_station_oldcode', 'Missing_records', 'Missing_percent']]
```

```python
num_excluded = excluded_stations['tele_station_oldcode'].nunique()
num_excluded
```

```python
remaining_stations = missing_by_station[
    missing_by_station['Missing_percent'] < 20
]
```

```python
num_remaining = remaining_stations['tele_station_oldcode'].nunique()
num_remaining
```

```python
summary_q4 = pd.DataFrame({
    'Category': ['Excluded stations (missing ≥ 20%)', 'Remaining stations'],
    'Number of stations': [num_excluded, num_remaining]
})

summary_q4
```

# **ข้อ5** เติมข้อมูลที่หายไปด้วยค่า Median ตาม Day of Year จาก historical แยกตามสถานี ตรวจสอบว่าไม่มี record หายอีก

```python
median_doy_station = (
    daily_rain_df
    .groupby(['station_code', 'doy'])['data']
    .median()
    .reset_index()
    .rename(columns={
        'station_code': 'tele_station_oldcode',
        'data': 'median_rainfall'
    })
)
```

```python
daily_input_df['date'] = pd.to_datetime(daily_input_df['measure_datetime'])
daily_input_df['doy'] = daily_input_df['date'].dt.dayofyear
```

```python
stations = remaining_stations['tele_station_oldcode'].unique()
all_dates = pd.date_range(
    daily_input_df['date'].min(),
    daily_input_df['date'].max(),
    freq='D'
)

full_index = pd.MultiIndex.from_product(
    [stations, all_dates],
    names=['tele_station_oldcode', 'date']
)

full_df = full_index.to_frame(index=False)
full_df['doy'] = full_df['date'].dt.dayofyear
```

```python
daily_complete = full_df.merge(
    daily_input_df,
    on=['tele_station_oldcode', 'date', 'doy'],
    how='left'
)
```

```python
daily_complete = daily_complete.merge(
    median_doy_station,
    on=['tele_station_oldcode', 'doy'],
    how='left'
)

daily_complete['rainfall_filled'] = daily_complete['rainfall_value']
daily_complete.loc[
    daily_complete['rainfall_filled'].isna(),
    'rainfall_filled'
] = daily_complete['median_rainfall']
```

```python
daily_complete['rainfall_filled'].isna().sum()
```

```python
pd.DataFrame({
    'Stage': ['Before filling', 'After filling'],
    'Missing records': [
        daily_complete['rainfall_value'].isna().sum(),
        daily_complete['rainfall_filled'].isna().sum()
    ]
})
```
